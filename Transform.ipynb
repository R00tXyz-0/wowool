```python
base_path = "othmane/othmane"

pipeline_description = [
    "english",
    "entity",
    "semantic-theme",
    "metadata",
    "business-news",
    {"name": "themes.app", "options": {"count": 1}},
    {
        "name": "dataframe.app",
        "options": {
            "annotations": [
                "Company",
                "CompanyPerformance", "CompanyAcquisition", "CompanyExpansion",
                "CompanyMerger", "MoneyAmount", "Date",
                "ProductLaunch", "ProductRecall", "CompanyPartnership",
                "CompanyLawsuit", "CompanyClosure", "CompanyDisruption",
                "CompanyWarning"
            ],
            "attributes": [
                "uri", "canonical","country",
                "sector", "themes", "sentence_id","polarity","doc_id"
            ],
            "metadata": [ "date", "document_url", "title"]
        }
    }
]

pipeline = Pipeline(pipeline_description)

files = list(Document.glob(base_path, "**/*.txt"))

print("Files found:", len(files))

all_rows = []


def process_document(document):
    print("Processing:", document.id)
    return pipeline(document)


with ThreadPoolExecutor() as executor:
    for doc in executor.map(process_document, files):
        all_rows.extend(doc.dataframe.rows)
        pass

df = pd.DataFrame(all_rows)
print(df.head(100))

df.to_csv("business-news-caro.csv", index=False, encoding="utf-8")
```
