```python
from pathlib import Path
import pandas as pd

base_path = Path("bn-dataset/09")

pipeline_description = [
    "english",
    "entity",
    "semantic-theme",
    "business-news",
    {
        "name": "themes.app",
        "options": {"count": 1}
    },
    {
        "name": "dataframe.app",
        "options": {
            "annotations": [
                "Person", "Organization", "Company", "Country", "Event",
                "CompanyPerformance", "CompanyAcquisition", "CompanyExpansion",
                "CompanyMerger", "MoneyAmount", "StockIndex", "StockExchange",
                "Position", "ProductLaunch", "ProductRecall", "CompanyPartnership",
                "CompanyLawsuit", "CompanyClosure", "CompanyDisruption", "CompanyWarning"
            ],
            "attributes": [
                "uri", "canonical", "gender", "country", "sector", "themes", "sent_id"
            ]
        }
    }
]

pipeline = Pipeline(pipeline_description)
all_rows = []

files = list(base_path.glob("*/*.txt"))



print("Files found:", len(files))

for ip in files:
    doc = pipeline(ip.data) if hasattr(ip, "data") else pipeline(ip.read_text(encoding="utf-8"))
    
    rows = doc.dataframe.rows

    for row in rows:
        if isinstance(row, dict):
            row["source_file"] = getattr(ip, "name", str(ip))

    all_rows.extend(rows)

df = pd.DataFrame(all_rows)

print("Total rows:", len(df))
display(df)
```

```python
output_path = "business-news01.csv"
df.to_csv(output_path, index=False, encoding="utf-8-sig")
print("CSV saved successfully:", output_path)
```
