```python
import pandas as pd
from pathlib import Path
from wowool.sdk import Pipeline
from wowool.document import Document
```

```python
import os
from getpass import getpass

os.environ["WOWOOL_SDK_KEY"] = getpass("Enter Wowool SDK Key: ")
```

```python
base_path = "bn-dataset/09"

pipeline_description = [
    "english",
    "entity",
    "semantic-theme",
    "business-news",
    {"name": "themes.app", "options": {"count": 1}},
    {
        "name": "dataframe.app",
        "options": {
            "annotations": [
                "Person", "Organization", "Company", "Country", "Event",
                "CompanyPerformance", "CompanyAcquisition", "CompanyExpansion",
                "CompanyMerger", "MoneyAmount", "StockIndex", "StockExchange",
                "Position", "ProductLaunch", "ProductRecall", "CompanyPartnership",
                "CompanyLawsuit", "CompanyClosure", "CompanyDisruption",
                "CompanyWarning"
            ],
            "attributes": [
                "uri", "canonical", "gender", "country",
                "sector", "themes", "sent_id"
            ]
        }
    }
]

pipeline = Pipeline(pipeline_description)

files = list(Document.glob(base_path, "**/*.txt"))

print("Files found:", len(files))

all_rows = []

for ip in files:
    doc = pipeline(ip)
    all_rows.extend(doc.dataframe.rows)

df = pd.DataFrame(all_rows)

print("Total rows:", len(df))
```

```python
all_rows = []

for document_id, ip in enumerate(files, start=1):
    doc = pipeline(ip)

    for row in doc.dataframe.rows:
        row["document_id"] = document_id
        all_rows.append(row)

df = pd.DataFrame(all_rows)

print("Total rows:", len(df))
df.head(20)
```

```python
output_path = "business-news01.csv"
df.to_csv(output_path, index=False, encoding="utf-8-sig")
print("CSV saved successfully:", output_path)
```
